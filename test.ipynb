```python
import numpy as np
from random import sample

import dask.array as da
```

```python
#pretty big array,choose chunk size automatically
my = da.random.uniform(0,10,3_000_000_000)
my
```

```python
#add some NaN to the array, Dask is able to handle the NaN's
indx = sample(range(0, 50_000_000), 400)
my[indx] =np.NaN
```

```python
from dask.distributed import Client,LocalCluster
cluster = LocalCluster()

client = Client(cluster)

client
```

```python
%%time
h, bins = da.histogram(my, bins=10, range=[0, 10])
hist = h.compute()
```

```python
sum(hist)
```
